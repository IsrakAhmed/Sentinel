```python
import os
from moviepy import VideoFileClip

# Root directory containing the 4 folders
root_dir = './multi_class__short_ds_resized_128'  # Replace with your actual path

# Supported video extensions
video_extensions = ('.mp4', '.avi', '.mov', '.mkv', '.flv', '.wmv')

# Total duration in seconds
total_duration = 0

# Walk through each subfolder and get all video files
for subfolder in os.listdir(root_dir):
    subfolder_path = os.path.join(root_dir, subfolder)
    
    if os.path.isdir(subfolder_path):
        for file in os.listdir(subfolder_path):
            if file.lower().endswith(video_extensions):
                video_path = os.path.join(subfolder_path, file)
                
                try:
                    clip = VideoFileClip(video_path)
                    duration = clip.duration  # duration in seconds
                    total_duration += duration
                    clip.close()
                except Exception as e:
                    print(f"Error reading {video_path}: {e}")

# Convert total duration to hours, minutes, and seconds
hours = int(total_duration // 3600)
minutes = int((total_duration % 3600) // 60)
seconds = int(total_duration % 60)

print(f"Total video duration: {hours}h {minutes}m {seconds}s")
```
