```python
import os
import cv2 as cv

# Define the source and destination directories
SOURCE_DIR = "short_ds"
DEST_DIR = "short_ds_resized_128"

# Target dimensions for resizing the videos
TARGET_WIDTH, TARGET_HEIGHT = 128, 128

def resize_video(video_path, output_path, target_width, target_height):
    # Capture the video
    video_reader = cv.VideoCapture(video_path)
    
    # Get original dimensions of the video
    original_width = int(video_reader.get(cv.CAP_PROP_FRAME_WIDTH))
    original_height = int(video_reader.get(cv.CAP_PROP_FRAME_HEIGHT))
    
    # If the video is already the target size, simply copy it
    if original_width == target_width and original_height == target_height:
        print(f"Video already in {target_width}x{target_height}, copying without resizing: {video_path}")
        os.system(f'cp "{video_path}" "{output_path}"')  # Copy the file
        return
    
    # Create a VideoWriter object to write the resized video
    fourcc = cv.VideoWriter_fourcc(*'mp4v')  # Codec for the output video
    fps = video_reader.get(cv.CAP_PROP_FPS)  # Original frame rate
    video_writer = cv.VideoWriter(output_path, fourcc, fps, (target_width, target_height))

    # Read frames, resize, and write to the new video
    while True:
        success, frame = video_reader.read()
        if not success:
            break
        resized_frame = cv.resize(frame, (target_width, target_height))
        video_writer.write(resized_frame)

    # Release resources
    video_reader.release()
    video_writer.release()
    print(f"Video resized and saved: {output_path}")

def process_videos(source_dir, dest_dir, target_width, target_height):
    # Create the destination directory if it doesn't exist
    if not os.path.exists(dest_dir):
        os.makedirs(dest_dir)

    # Iterate over each class folder inside the source directory
    for class_name in os.listdir(source_dir):
        class_path = os.path.join(source_dir, class_name)
        
        # Check if it's a directory
        if os.path.isdir(class_path):
            # Create corresponding resized folder
            resized_class_path = os.path.join(dest_dir, f"{class_name}_resized")
            if not os.path.exists(resized_class_path):
                os.makedirs(resized_class_path)

            # Iterate over all video files in the class folder
            for video_file in os.listdir(class_path):
                video_path = os.path.join(class_path, video_file)
                if video_file.endswith(('.mp4', '.avi', '.mov')):  # Filter by video formats
                    resized_video_path = os.path.join(resized_class_path, video_file)

                    # Call the resize function for each video
                    resize_video(video_path, resized_video_path, target_width, target_height)

# Call the function to process and resize videos
process_videos(SOURCE_DIR, DEST_DIR, TARGET_WIDTH, TARGET_HEIGHT)
```
